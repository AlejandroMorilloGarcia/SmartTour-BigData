## Parte 1. Principio multiplicativo

### 1.1 Calcular cuanto outfit diferentes hay

Calculo la cantidad de combiniaciones de outfits que se pueden hacer 

```python
def calcular_outfits(**prendas):
    total = 1
    for categoria, cantidad in prendas.items():
        total *= cantidad
    return total

camisetas = ("C1", "C2", "C3")
pantalones = ("P1", "P1")
zapatos = ("Z1", "Z2")

print(f"Cantidad de combinaciones de outfits: {calcular_outfits(camisetas=len(camisetas), pantalones=len(pantalones), zapatos=len(zapatos))}")
```

## Parte 2. Generación de combinaciones

### 2.1 Generar todas la combinaciones posibles y representarlo en una lista o tabla

```python
from itertools import product
outfits = list(product(camisetas, pantalones, zapatos))
c = 0
for o in outfits:
    c += 1
    print(f"{c}: {o[0]}, {o[1]}, {o[2]}")
    
```
