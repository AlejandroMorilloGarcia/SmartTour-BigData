Parte 1. Construcción de los conjuntos

```python
import os
import pandas as pd
jugadores_ClubA = set()
jugadores_ClubB = set()

#cargar jugadores clubA
for archivo in os.listdir("../../datasets/dataFutbol/club_a"):
    ruta = os.path.join("../../datasets/dataFutbol/club_a", archivo)
    df = pd.read_csv(ruta)
    jugadores_ClubA.update(df["jugador"].dropna().astype(str))

#cargar jugadores clubB
for archivo in os.listdir("../../datasets/dataFutbol/club_b"):
    ruta = os.path.join("../../datasets/dataFutbol/club_b", archivo)
    df = pd.read_csv(ruta)
    jugadores_ClubB.update(df["jugador"].dropna().astype(str))
print(jugadores_ClubA)
print("--------------------------------------------------------------")
print(jugadores_ClubB)
```

Parte 2. Operaciones con conjuntos

2.1 Jugadores diferentes que han pasado por alguno de los clubes

```python
union = jugadores_ClubA | jugadores_ClubB
print(f"jugadores que han pasado por alguno de los clubes: {union}")
```

2.2 Jugadores que han pertenecido a ambos equipos

```python
interseccion = jugadores_ClubA & jugadores_ClubB
print(f"jugadores que han a ambos equipos: {interseccion}")
```

2.3 Jugadores que han jugado solo en el club A

```python
diferencia = jugadores_ClubA - jugadores_ClubB
print(f"jugadores que han jugado en el club A: {diferencia}")
```

2.4 Juadores que solo han jugado exclusivamente en uno de los dos equipos

```python
diferenciaSimetrica = jugadores_ClubA ^ jugadores_ClubB
print(f"jugadores que han jugado exclusivamente en uno de los dos equipos: {diferenciaSimetrica}")
```

Parte 3. Análisis de resultados

3.1 porcentaje de jugadores que han jugado en ambos clubs

```python
porcentaje = len(interseccion) / len(union) * 100
print(f"El {porcentaje:.2f}% de los jugadores del universo ha jugado en ambos clubes.")
```

3.2 Club con mayores jugadores exclusivos

```python
exclusivosA = jugadores_ClubA - jugadores_ClubB
exclusivosB = jugadores_ClubB - jugadores_ClubA
if len(exclusivosA) > len(exclusivosB):
    print("El Club A ha tenido más jugadores exclusivos.")
elif len(exclusivosB) > len(exclusivosA):
    print("El Club B ha tenido más jugadores exclusivos.")
else:
    print("Ambos clubes han tenido el mismo número de jugadores exclusivos.")
```
